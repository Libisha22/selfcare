```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

df = pd.read_csv("diabetes.csv")

print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[cols] = df[cols].replace(0, np.nan)

df.fillna(df.median(), inplace=True)
df.drop_duplicates(inplace=True)

print("\nAfter Cleaning:")
print(df.isnull().sum())

plt.figure()
sns.countplot(x='Outcome', data=df)
plt.title("Diabetes Outcome Distribution")
plt.show()

plt.figure()
sns.histplot(df['Glucose'], kde=True)
plt.title("Glucose Level Distribution")
plt.show()

plt.figure()
sns.boxplot(x='Outcome', y='BMI', data=df)
plt.title("BMI vs Diabetes Outcome")
plt.show()

plt.figure()
sns.scatterplot(x='Age', y='Glucose', hue='Outcome', data=df)
plt.title("Age vs Glucose")
plt.show()

plt.figure()
sns.heatmap(df.corr(), annot=True)
plt.title("Feature Correlation Heatmap")
plt.show()

grouped = df.groupby('Outcome').mean()
print("\nFeature Means by Outcome:")
print(grouped)

print("\nKey Insights:")
print("- Glucose has strong influence on diabetes.")
print("- Higher BMI increases risk.")
print("- Older age group shows higher probability.")
print("- Multiple features together improve prediction accuracy.")
```
